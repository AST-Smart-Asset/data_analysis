```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
from google.colab import files
import warnings

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("=" * 70)
print("DATASET OVERVIEW")
print("=" * 70)

print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]:,}")
print(f"Memory Usage: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

display(df.head())
display(df.tail())

print("\nDATA TYPES")
display(df.dtypes.to_frame("Data Type"))

print("\nMISSING VALUES BEFORE CLEANING")
missing_before = df.isnull().sum().to_frame("Missing Values")
missing_before["Percentage"] = (
    missing_before["Missing Values"] / len(df) * 100
).round(2)
display(missing_before[missing_before["Missing Values"] > 0])

print("\nDUPLICATE ROWS BEFORE CLEANING")
print(df.duplicated().sum())

df["snapshot_date"] = pd.to_datetime(
    df["snapshot_date"], errors="coerce"
)

df = df.drop_duplicates().copy()

df["department"] = df["department"].fillna("Unknown")
df["building"] = df["building"].fillna("Unknown")
df["condition_at_snapshot"] = df["condition_at_snapshot"].fillna("Unknown")
df["warranty_active"] = df["warranty_active"].fillna("Unknown")
df["location_type"] = df["location_type"].fillna("Unknown")
df["category_code"] = df["category_code"].fillna("Unknown")
df["category_group_code"] = df["category_group_code"].fillna("Unknown")

numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

for column in numeric_columns:
    if column not in ["failure_within_30d", "days_to_next_failure"]:
        df[column] = df[column].fillna(df[column].median())

df["missing_days_to_failure"] = (
    df["days_to_next_failure"].isna().astype(int)
)

df["failure_status"] = df["failure_within_30d"].map({
    0: "No Failure",
    1: "Failure"
})

df["age_years"] = df["age_days"] / 365.25

df["life_used_percentage"] = np.where(
    df["useful_life_months"] > 0,
    (df["age_months"] / df["useful_life_months"]) * 100,
    np.nan
)

df["total_maintenance_events"] = (
    df["prior_work_orders_count"] +
    df["maintenance_last_90d_count"]
)

df["maintenance_cost_per_work_order"] = np.where(
    df["prior_work_orders_count"] > 0,
    df["prior_maintenance_cost_total"] /
    df["prior_work_orders_count"],
    0
)

df["failure_history_flag"] = (
    df["prior_failures_count"] > 0
).astype(int)

df["high_maintenance_flag"] = (
    df["maintenance_last_90d_count"] >
    df["maintenance_last_90d_count"].median()
).astype(int)

df["warranty_status"] = df["warranty_active"].astype(str).str.lower().map({
    "t": "Active",
    "true": "Active",
    "1": "Active",
    "y": "Active",
    "yes": "Active",
    "f": "Inactive",
    "false": "Inactive",
    "0": "Inactive",
    "n": "Inactive",
    "no": "Inactive"
}).fillna("Unknown")

df["snapshot_year"] = df["snapshot_date"].dt.year
df["snapshot_month"] = df["snapshot_date"].dt.month
df["snapshot_month_name"] = df["snapshot_date"].dt.month_name()

df["age_group"] = pd.cut(
    df["age_years"],
    bins=[-np.inf, 1, 3, 5, 10, np.inf],
    labels=["0-1 Years", "1-3 Years", "3-5 Years", "5-10 Years", "10+ Years"]
)

df["cost_group"] = pd.qcut(
    df["purchase_cost"],
    q=4,
    labels=["Low Cost", "Medium Cost", "High Cost", "Very High Cost"],
    duplicates="drop"
)

print("\n" + "=" * 70)
print("DATA CLEANING RESULTS")
print("=" * 70)

print(f"Rows After Cleaning: {df.shape[0]:,}")
print(f"Columns After Preparation: {df.shape[1]:,}")
print(f"Duplicate Rows After Cleaning: {df.duplicated().sum()}")

missing_after = df.isnull().sum().to_frame("Missing Values")
missing_after["Percentage"] = (
    missing_after["Missing Values"] / len(df) * 100
).round(2)

display(missing_after[missing_after["Missing Values"] > 0])

print("\nDESCRIPTIVE STATISTICS")
display(df.describe(include="all").T)

print("\n" + "=" * 70)
print("EXPLORATORY DATA ANALYSIS")
print("=" * 70)

print("\nFAILURE DISTRIBUTION")

failure_counts = df["failure_status"].value_counts()
failure_percentages = (
    df["failure_status"].value_counts(normalize=True) * 100
).round(2)

failure_summary = pd.DataFrame({
    "Count": failure_counts,
    "Percentage": failure_percentages
})

display(failure_summary)

print("\nNUMERIC CORRELATIONS WITH FAILURE TARGET")

numeric_df = df.select_dtypes(include=np.number)

correlation_with_target = (
    numeric_df.corr(numeric_only=True)["failure_within_30d"]
    .sort_values(ascending=False)
    .to_frame("Correlation")
)

display(correlation_with_target)

print("\nFAILURE RATE BY CATEGORY")

category_failure = (
    df.groupby("category_group_code", observed=True)
    .agg(
        Total_Assets=("asset_tag", "count"),
        Failures=("failure_within_30d", "sum"),
        Failure_Rate=("failure_within_30d", "mean"),
        Average_Cost=("purchase_cost", "mean"),
        Average_Age=("age_years", "mean")
    )
    .sort_values("Failure_Rate", ascending=False)
)

category_failure["Failure_Rate"] *= 100
display(category_failure)

print("\nFAILURE RATE BY CONDITION")

condition_failure = (
    df.groupby("condition_at_snapshot", observed=True)
    .agg(
        Total_Assets=("asset_tag", "count"),
        Failures=("failure_within_30d", "sum"),
        Failure_Rate=("failure_within_30d", "mean"),
        Average_Cost=("purchase_cost", "mean")
    )
    .sort_values("Failure_Rate", ascending=False)
)

condition_failure["Failure_Rate"] *= 100
display(condition_failure)

print("\nFAILURE RATE BY WARRANTY STATUS")

warranty_failure = (
    df.groupby("warranty_status", observed=True)
    .agg(
        Total_Assets=("asset_tag", "count"),
        Failures=("failure_within_30d", "sum"),
        Failure_Rate=("failure_within_30d", "mean")
    )
    .sort_values("Failure_Rate", ascending=False)
)

warranty_failure["Failure_Rate"] *= 100
display(warranty_failure)

print("\nFAILURE RATE BY AGE GROUP")

age_failure = (
    df.groupby("age_group", observed=True)
    .agg(
        Total_Assets=("asset_tag", "count"),
        Failures=("failure_within_30d", "sum"),
        Failure_Rate=("failure_within_30d", "mean")
    )
)

age_failure["Failure_Rate"] *= 100
display(age_failure)

print("\nFAILURE RATE BY BUILDING")

building_failure = (
    df.groupby("building", observed=True)
    .agg(
        Total_Assets=("asset_tag", "count"),
        Failures=("failure_within_30d", "sum"),
        Failure_Rate=("failure_within_30d", "mean")
    )
    .sort_values("Failure_Rate", ascending=False)
)

building_failure["Failure_Rate"] *= 100
display(building_failure)

print("\nFAILURE RATE BY LOCATION TYPE")

location_failure = (
    df.groupby("location_type", observed=True)
    .agg(
        Total_Assets=("asset_tag", "count"),
        Failures=("failure_within_30d", "sum"),
        Failure_Rate=("failure_within_30d", "mean")
    )
    .sort_values("Failure_Rate", ascending=False)
)

location_failure["Failure_Rate"] *= 100
display(location_failure)

print("\n" + "=" * 70)
print("DATA VISUALIZATION")
print("=" * 70)

plt.figure(figsize=(10, 6))
sns.countplot(
    data=df,
    x="failure_status",
    order=df["failure_status"].value_counts().index
)
plt.title("Failure Status Distribution")
plt.xlabel("Failure Status")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
failure_rate_category = category_failure["Failure_Rate"].sort_values(
    ascending=False
)
sns.barplot(
    x=failure_rate_category.values,
    y=failure_rate_category.index
)
plt.title("Failure Rate by Category Group")
plt.xlabel("Failure Rate (%)")
plt.ylabel("Category Group")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.histplot(
    data=df,
    x="age_years",
    hue="failure_status",
    bins=30,
    kde=True,
    element="step"
)
plt.title("Asset Age Distribution by Failure Status")
plt.xlabel("Asset Age (Years)")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=df,
    x="failure_status",
    y="age_years"
)
plt.title("Asset Age vs Failure Status")
plt.xlabel("Failure Status")
plt.ylabel("Age (Years)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=df,
    x="failure_status",
    y="purchase_cost"
)
plt.title("Purchase Cost vs Failure Status")
plt.xlabel("Failure Status")
plt.ylabel("Purchase Cost")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=df,
    x="failure_status",
    y="prior_failures_count"
)
plt.title("Previous Failures vs Failure Status")
plt.xlabel("Failure Status")
plt.ylabel("Previous Failures Count")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=df,
    x="failure_status",
    y="maintenance_last_90d_count"
)
plt.title("Recent Maintenance vs Failure Status")
plt.xlabel("Failure Status")
plt.ylabel("Maintenance Count in Last 90 Days")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.scatterplot(
    data=df,
    x="age_years",
    y="purchase_cost",
    hue="failure_status",
    alpha=0.6
)
plt.title("Asset Age vs Purchase Cost")
plt.xlabel("Age (Years)")
plt.ylabel("Purchase Cost")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.histplot(
    data=df,
    x="purchase_cost",
    bins=40,
    kde=True
)
plt.title("Purchase Cost Distribution")
plt.xlabel("Purchase Cost")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.histplot(
    data=df,
    x="prior_work_orders_count",
    bins=30,
    kde=True
)
plt.title("Previous Work Orders Distribution")
plt.xlabel("Previous Work Orders")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(
    data=df,
    x="warranty_status",
    y="failure_within_30d",
    estimator=np.mean
)
plt.title("Failure Rate by Warranty Status")
plt.xlabel("Warranty Status")
plt.ylabel("Failure Rate")
plt.gca().yaxis.set_major_formatter(
    plt.matplotlib.ticker.PercentFormatter(1.0)
)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(
    data=df,
    x="age_group",
    y="failure_within_30d",
    estimator=np.mean
)
plt.title("Failure Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Failure Rate")
plt.gca().yaxis.set_major_formatter(
    plt.matplotlib.ticker.PercentFormatter(1.0)
)
plt.tight_layout()
plt.show()

plt.figure(figsize=(14, 8))
correlation_matrix = df.select_dtypes(
    include=np.number
).corr()

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
monthly_failure = (
    df.groupby("snapshot_month_name", observed=True)["failure_within_30d"]
    .mean()
    .reindex([
        "January", "February", "March", "April",
        "May", "June", "July", "August",
        "September", "October", "November", "December"
    ])
)

sns.lineplot(
    x=monthly_failure.index,
    y=monthly_failure.values,
    marker="o"
)
plt.title("Failure Rate by Snapshot Month")
plt.xlabel("Month")
plt.ylabel("Failure Rate")
plt.xticks(rotation=45)
plt.gca().yaxis.set_major_formatter(
    plt.matplotlib.ticker.PercentFormatter(1.0)
)
plt.tight_layout()
plt.show()

print("\n" + "=" * 70)
print("INSIGHT GENERATION")
print("=" * 70)

total_records = len(df)
total_failures = int(df["failure_within_30d"].sum())
overall_failure_rate = df["failure_within_30d"].mean() * 100

print(f"Total Records: {total_records:,}")
print(f"Total Failures: {total_failures:,}")
print(f"Overall Failure Rate: {overall_failure_rate:.2f}%")

if len(category_failure) > 0:
    highest_category = category_failure["Failure_Rate"].idxmax()
    highest_category_rate = category_failure["Failure_Rate"].max()
    highest_category_count = category_failure.loc[
        highest_category, "Total_Assets"
    ]

    print(
        f"Highest Category Failure Rate: {highest_category} "
        f"({highest_category_rate:.2f}%, "
        f"{highest_category_count:,} records)"
    )

if len(condition_failure) > 0:
    highest_condition = condition_failure["Failure_Rate"].idxmax()
    highest_condition_rate = condition_failure["Failure_Rate"].max()

    print(
        f"Highest Condition Failure Rate: {highest_condition} "
        f"({highest_condition_rate:.2f}%)"
    )

if len(age_failure) > 0:
    highest_age_group = age_failure["Failure_Rate"].idxmax()
    highest_age_rate = age_failure["Failure_Rate"].max()

    print(
        f"Highest Age Group Failure Rate: {highest_age_group} "
        f"({highest_age_rate:.2f}%)"
    )

if len(warranty_failure) > 0:
    highest_warranty = warranty_failure["Failure_Rate"].idxmax()
    highest_warranty_rate = warranty_failure["Failure_Rate"].max()

    print(
        f"Highest Warranty Status Failure Rate: {highest_warranty} "
        f"({highest_warranty_rate:.2f}%)"
    )

if len(building_failure) > 0:
    highest_building = building_failure["Failure_Rate"].idxmax()
    highest_building_rate = building_failure["Failure_Rate"].max()

    print(
        f"Highest Building Failure Rate: {highest_building} "
        f"({highest_building_rate:.2f}%)"
    )

if len(location_failure) > 0:
    highest_location = location_failure["Failure_Rate"].idxmax()
    highest_location_rate = location_failure["Failure_Rate"].max()

    print(
        f"Highest Location Type Failure Rate: {highest_location} "
        f"({highest_location_rate:.2f}%)"
    )

if "prior_failures_count" in df.columns:
    failure_history_comparison = (
        df.groupby("failure_history_flag")["failure_within_30d"]
        .mean() * 100
    )

    print("\nFailure Rate by Previous Failure History")
    display(failure_history_comparison.to_frame("Failure Rate (%)"))

if "high_maintenance_flag" in df.columns:
    maintenance_comparison = (
        df.groupby("high_maintenance_flag")["failure_within_30d"]
        .mean() * 100
    )

    print("\nFailure Rate by Maintenance Activity")
    display(maintenance_comparison.to_frame("Failure Rate (%)"))

print("\nTOP NUMERIC FEATURES CORRELATED WITH FAILURE")

display(
    correlation_with_target
    .drop(index="failure_within_30d", errors="ignore")
    .head(10)
)

print("\n" + "=" * 70)
print("CLEANED DATASET EXPORT")
print("=" * 70)

output_file = "cleaned_ml_training_snapshots.csv"

df.to_csv(output_file, index=False)

print(f"Exported File: {output_file}")
print(f"Final Shape: {df.shape}")

files.download(output_file)

print("\nPROJECT COMPLETED SUCCESSFULLY")
```
